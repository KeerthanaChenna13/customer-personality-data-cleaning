```python
df = pd.read_csv('marketing_campaign.csv', sep='\t')
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# Calculate median income
median_income = df['Income'].median()

# Fill missing income values
df['Income'] = df['Income'].fillna(median_income)

# Check missing values again
print(df.isnull().sum())
```

```python
print("Total duplicate rows:", df.duplicated().sum())
```

```python
df['Dt_Customer'] = pd.to_datetime(
    df['Dt_Customer'],
    format='%d-%m-%Y',
    errors='coerce'
)

print(df['Dt_Customer'].dtype)
```

```python
# Check unique values in Education
print("Education values:")
print(df['Education'].value_counts())

# Check unique values in Marital_Status
print("\nMarital Status values:")
print(df['Marital_Status'].value_counts())
```

```python
df['Education'] = df['Education'].str.strip()

df['Marital_Status'] = df['Marital_Status'].str.strip()
```

```python
print(df[['Year_Birth', 'Income']].describe())
```

```python
# Find unusually old birth years
print("Suspicious birth year records:")
print(df[df['Year_Birth'] < 1930][['ID', 'Year_Birth']])

# Find unusually high income records
print("\nSuspicious income records:")
print(df[df['Income'] > 150000][['ID', 'Income']])
```

```python
Q1 = df['Income'].quantile(0.25)
Q3 = df['Income'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = df[
    (df['Income'] < lower_limit) |
    (df['Income'] > upper_limit)
]

print("Number of Income Outliers:", len(outliers))
```

```python
# Display complete details of suspicious records

print(df[df['Year_Birth'] < 1930])

print("\nIncome outlier details:")

print(df[df['Income'] > 117416.25])
```

```python
# Identify extreme income value

df[df['Income'] == 666666]
```

```python
# Replace the suspicious income value with median income

median_income = df['Income'].median()

df.loc[df['Income'] == 666666, 'Income'] = median_income

print(df[df['ID'] == 9432][['ID', 'Income']])
```

```python
df['Age'] = 2014 - df['Year_Birth']

print(df['Age'].describe())
```

```python
print(df[df['Age'] > 90][['ID', 'Year_Birth', 'Age']])
```

```python
print("Dataset Shape:", df.shape)

print("Missing Values:", df.isnull().sum().sum())

print("Duplicate Records:", df.duplicated().sum())

print(df.info())
```

```python
# Convert column names to lowercase
# Replace spaces with underscores

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

print(df.columns.tolist())
```

```python
print(df.dtypes)
```

```python
print("Total Rows and Columns:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())
```

```python
df.to_csv('cleaned_customer_personality.csv', index=False)

print("Cleaned dataset exported successfully!")
```
